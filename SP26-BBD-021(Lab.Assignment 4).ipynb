```python
# 1
from datetime import datetime

def greet_with_time(name):
    current_time = datetime.now().strftime("%H:%M:%S")
    print(f"Hello {name}! Current time is {current_time}")

greet_with_time("Maham")
```

```python
#2
def calculate_average(x, y, z):
    average = (x + y + z) / 3
    return average

print(calculate_average(10, 20, 30))
```

```python
# 3
def calculate_discount(price, discount_percentage=10):
    discount = price * discount_percentage / 100
    final_price = price - discount
    return final_price

print(calculate_discount(1000))
print(calculate_discount(1000, 20))
```

```python
# 4
def print_info(name, age, city):
    print("Name:", name)
    print("Age:", age)
    print("City:", city)

print_info(name="Maham", age=19, city="Bahawalpur")
```

```python
# 5
def find_max(*numbers):
    return max(numbers)

print(find_max(10, 45, 2, 100, 34))
```

```python
# 6
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(6))
```

```python
# 7
def square(num):
    return num * num

def apply_function(func, numbers):
    result = []
    for num in numbers:
        result.append(func(num))
    return result

print(apply_function(square, [1, 2, 3, 4]))
```

```python
# 8
def double(num):
    return num * 2

def apply_operation(func, numbers):
    return [func(num) for num in numbers]

print(apply_operation(double, [4, 5, 6, 7]))
```

```python
# 9
import time

def timer_decorator(func):
    def wrapper():
        start = time.time()
        
        func()
        
        end = time.time()
        print("Execution Time:", end - start, "seconds")
    
    return wrapper

@timer_decorator
def sample_function():
    time.sleep(2)
    print("Function is running...")

sample_function()
```
